Given the head to a singly linked list, where each node also has a “random” pointer that points to anywhere in the linked list, deep clone the list.

```python
class Node:
  def __init__(self, val, ptr):
    self.next = None
    self.ptr = ptr
    self.val = val


def deepClone(head):
  if not head:
    return None

  new_head = Node(head.val, head.ptr)
  ptr1 = head.next
  old_new_map = {}
  #old_ptr_map = {}
  old_new_map[head] = new_head
  #old_ptr_map[head] =
  while ptr1:
    new_node = Node(ptr1.val, ptr1.ptr)
    old_new_map[ptr1] = new_node
    new_head.next = new_node
    new_head = new_node
    ptr1 = ptr1.next

  new_head = old_new_map[head]
  while new_head:
    ptr = new_head.ptr
    new_head.ptr = old_new_map[ptr]
    new_head = new_head.next

  return old_new_map[head]





```

```python
# Create a sample linked list with random pointers
node1 = Node(1, None)
node2 = Node(2, None)
node3 = Node(3, None)
node4 = Node(4, None)

node1.next = node2
node2.next = node3
node3.next = node4

node1.ptr = node3  # Random pointer from node1 to node3
node2.ptr = node1  # Random pointer from node2 to node1
node3.ptr = node4  # Random pointer from node3 to node4
node4.ptr = node2  # Random pointer from node4 to node2

head = node1

# Function to print the linked list (including random pointers)
def print_list(head):
    current = head
    while current:
        print(f"Value: {current.val}, Next: {current.next.val if current.next else 'None'}, Random: {current.ptr.val if current.ptr else 'None'}")
        current = current.next

print("Original list:")
print_list(head)

# Deep clone the list
cloned_head = deepClone(head)

print("\nCloned list:")
print_list(cloned_head)

# Verify the clone
# Check if the values are the same and the nodes are different objects
current_orig = head
current_cloned = cloned_head
is_deep_clone = True
while current_orig:
    if current_orig.val != current_cloned.val or current_orig is current_cloned:
        is_deep_clone = False
        break
    current_orig = current_orig.next
    current_cloned = current_cloned.next

# Check if random pointers point to the correct cloned nodes
current_orig = head
current_cloned = cloned_head
while current_orig:
    if (current_orig.ptr and not current_cloned.ptr) or (not current_orig.ptr and current_cloned.ptr) or \
       (current_orig.ptr and current_cloned.ptr and current_orig.ptr.val != current_cloned.ptr.val) or \
       (current_orig.ptr and current_cloned.ptr and current_orig.ptr is current_cloned.ptr):
        is_deep_clone = False
        break
    current_orig = current_orig.next
    current_cloned = current_cloned.next


print("\nIs the clone a deep clone?", is_deep_clone)
```
