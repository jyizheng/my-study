```python
## put/get/delete/get_all_keys


class Node:
    def __init__(self, key=None, value=None):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None


class KVStore:
    def __init__(self):
        self.head = Node()
        self.tail = Node()

        # circular
        self.head.next = self.tail
        self.head.prev = self.tail
        self.tail.next = self.head
        self.tail.prev = self.head

        self.map = {}

    def _delete_node(self, node: Node):
        """Remove node from list."""
        node.prev.next = node.next
        node.next.prev = node.prev
        #node.prev = node.next = None

    def _insert_before_tail(self, node: Node):
        """Insert node before tail (end of order)."""
        prev = self.tail.prev
        prev.next = node
        node.prev = prev
        node.next = self.tail
        self.tail.prev = node

    def put(self, key, value):
        """Insert or update key-value, move to end if exists."""
        if key in self.map:
            node = self.map[key]
            node.value = value
            # move node to end
            self._delete_node(node)
            self._insert_before_tail(node)
        else:
            node = Node(key, value)
            self.map[key] = node
            self._insert_before_tail(node)

    def get(self, key):
        """Return value if key exists, else None."""
        return self.map[key].value if key in self.map else None

    def delete(self, key):
        """Delete key if exists."""
        if key in self.map:
            node = self.map.pop(key)
            self._delete_node(node)

    def get_all_keys(self):
        """Return keys in insertion order."""
        keys = []
        cur = self.head.next
        while cur != self.tail:
            keys.append(cur.key)
            cur = cur.next
        return keys



kv = KVStore()
kv.put("a", 1)
kv.put("b", 2)
kv.put("c", 3)

print(kv.get("b"))        # 2
kv.delete("b")

print(kv.get_all_keys())  # ['a', 'c']

kv.put("a", 99)           # update value
print(kv.get("a"))        # 99



```
