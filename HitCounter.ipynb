Good morning! Here's your coding interview problem for today.

This question was asked by Riot Games.

Design and implement a HitCounter class that keeps track of requests (or hits). It should support the following operations:

record(timestamp): records a hit that happened at timestamp
total(): returns the total number of hits recorded
range(lower, upper): returns the number of hits that occurred between timestamps lower and upper (inclusive)
Follow-up: What if our system has limited memory?

```python
import bisect

class HitCounter:
    """
    A class to record and query hit counts based on timestamps.
    This implementation assumes unlimited memory.
    """
    def __init__(self):
        """Initializes the HitCounter."""
        self._hits = []

    def record(self, timestamp: int) -> None:
        """
        Records a hit at a given timestamp.

        Time Complexity: O(1) - Assuming timestamps are sequential.
                         If timestamps can be out of order, we would use
                         bisect.insort(), making it O(log N).
        Space Complexity: O(N) - Where N is the total number of hits.
        """
        # Assumes timestamps arrive in increasing order for O(1) append.
        self._hits.append(timestamp)

    def total(self) -> int:
        """
        Returns the total number of hits recorded.

        Time Complexity: O(1)
        """
        return len(self._hits)

    def range(self, lower: int, upper: int) -> int:
        """
        Returns the number of hits between two timestamps (inclusive).

        Time Complexity: O(log N) - Due to binary search.
        """
        if lower > upper:
            return 0

        # Find the insertion point for `lower`, which is the index of the
        # first element greater than or equal to `lower`.
        start_index = bisect.bisect_left(self._hits, lower)

        # Find the insertion point for `upper`, which is the index of the
        # first element strictly greater than `upper`.
        end_index = bisect.bisect_right(self._hits, upper)

        return end_index - start_index

```

```python
import unittest

class TestHitCounter(unittest.TestCase):
    def test_record_and_total(self):
        counter = HitCounter()
        self.assertEqual(counter.total(), 0)
        counter.record(1)
        self.assertEqual(counter.total(), 1)
        counter.record(2)
        counter.record(3)
        self.assertEqual(counter.total(), 3)

    def test_range(self):
        counter = HitCounter()
        counter.record(1)
        counter.record(2)
        counter.record(3)
        counter.record(5)
        counter.record(8)
        counter.record(10)
        counter.record(10)

        self.assertEqual(counter.range(2, 8), 4) # hits at 2, 3, 5, 8
        self.assertEqual(counter.range(0, 1), 1) # hit at 1
        self.assertEqual(counter.range(10, 20), 2) # hits at 10, 10
        self.assertEqual(counter.range(7, 9), 1) # hit at 8
        self.assertEqual(counter.range(11, 15), 0) # no hits
        self.assertEqual(counter.range(3, 3), 1) # hit at 3
        self.assertEqual(counter.range(5, 2), 0) # invalid range

    def test_empty_range(self):
        counter = HitCounter()
        self.assertEqual(counter.range(1, 10), 0)

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
