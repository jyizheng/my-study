```python
class Solution:
    def process_one_path(self, path, my_dict):
        paths = path.split(" ")
        dir_path = paths[0]
        file_paths = paths[1:]

        for f in file_paths:
            name_and_data = f.split("(")
            file_name = name_and_data[0]
            file_data = name_and_data[1][:-1]
            file_name = dir_path + "/" + file_name
            lst = my_dict.get(file_data, [])
            lst.append(file_name)
            my_dict[file_data] = lst


    def findDuplicate(self, paths: List[str]) -> List[List[str]]:
        mydict = {}
        for path in paths:
            self.process_one_path(path, mydict)

        res = []
        for k, v in mydict.items():
            if len(v) > 1:
                res.append(v)

        return res


```
