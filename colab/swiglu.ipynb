```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class SwiGLUFeedForward(nn.Module):
  def __init__(self, d_model: int, d_ff: int):
    super().__init__()
    # TODO: 初始化SwiGLU所需的三个线性层
    # --- Solution Start ---
    self.w1 = nn.Linear(d_model, d_ff, bias=False)
    self.w2 = nn.Linear(d_model, d_ff, bias=False)
    self.w3 = nn.Linear(d_ff, d_model, bias=False)
    # --- Solution End ---

  def forward(self, x: torch.Tensor):
    # TODO: 实现SwiGLU的前向传播逻辑
    # Swish(W1x) * W2x，然后通过W3
    # --- Solution Start ---
    gate = self.w1(x)
    up = self.w2(x)
    return self.w3(F.silu(gate) * up)
    # --- Solution End ---
```

```python
def run_swiglu_verification():
  print("="*30 + "\n6. 验证 SwiGLU 实现...\n" + "="*30)
  d_model, d_ff, b, s = 128, 256, 4, 16

  # 1. 初始化检查
  try:
    model = SwiGLUFeedForward(d_model, d_ff)
    print("[检查点 1/3] SwiGLU 实例化成功。")
  except Exception as e:
    print(f"[检查点 1/3] SwiGLU 实例化失败: {e}"); return

  # 2. 前向传播和输出形状检查
  try:
    inp = torch.randn(b, s, d_model)
    out = model(inp)
    if out.shape == inp.shape:
      print(f"[检查点 2/3] 前向传播成功，输出形状正确: {out.shape}。")
    else:
      print(f"[检查点 2/3] 输出形状错误！"); return
  except Exception as e:
    print(f"[检查点 2/3] 前向传播失败: {e}"); return

  # 3. 梯度检查
  try:
    out.sum().backward()
    print("[检查点 3/3] 梯度计算成功。")
  except Exception as e:
    print(f"[检查点 3/3] 梯度计算失败: {e}"); return

  print("SwiGLU 验证通过!\n")

if __name__ == '__main__':
  run_swiglu_verification()
```
