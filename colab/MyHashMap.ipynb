```python
class Node:
    def __init__(self, key:int, value:int):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None


class MyHashMap:

    def __init__(self):
        self.dict = {}
        self.head = None

    def put(self, key: int, value: int) -> None:
        if not key in self.dict:
            node = Node(key, value)
            if not self.head:
                self.head = node
                node.prev = node
                node.next = node
            else:
                prev = self.head.prev
                prev.next = node
                node.prev = prev
                node.next = self.head
                self.head.prev = node

            self.dict[key] = node
        else:
            node = self.dict.get(key)
            node.value = value

    def get(self, key: int) -> int:
        if not key in self.dict:
            return -1
        else:
            node = self.dict.get(key)
            return node.value
    def remove(self, key: int) -> None:
        if key in self.dict:
            node = self.dict.get(key)
            if node == self.head:
                if self.head == self.head.next:
                    self.head = None
                else:
                    self.head = self.head.next
            else:
                prev = node.prev
                next = node.next
                prev.next = next
                next.prev = prev

            del self.dict[key]




# Your MyHashMap object will be instantiated and called as such:
# obj = MyHashMap()
# obj.put(key,value)
# param_2 = obj.get(key)
# obj.remove(key)
```
