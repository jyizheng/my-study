```python
class MyCircularQueue:

    def __init__(self, k: int):
        self.data = [0]*k
        self.head = -1
        self.tail = -1
        self.size = 0
        self.k = k

    def enQueue(self, value: int) -> bool:
        if self.size == self.k:
            return False
        else:
            self.tail += 1
            pos = self.tail % self.k
            self.data[pos] = value
            self.size += 1
            if self.head == -1:
                self.head = 0
            return True
    def deQueue(self) -> bool:
        if self.size == 0:
            return False
        self.head += 1
        pos = self.head % self.k
        self.size -= 1
        return True

    def Front(self) -> int:
        if self.size == 0:
            return -1
        else:
            return self.data[self.head % self.k]

    def Rear(self) -> int:
        if self.size == 0:
            return -1
        else:
            return self.data[self.tail % self.k]

    def isEmpty(self) -> bool:
        return self.size == 0

    def isFull(self) -> bool:
        return self.size == self.k



# Your MyCircularQueue object will be instantiated and called as such:
# obj = MyCircularQueue(k)
# param_1 = obj.enQueue(value)
# param_2 = obj.deQueue()
# param_3 = obj.Front()
# param_4 = obj.Rear()
# param_5 = obj.isEmpty()
# param_6 = obj.isFull()
```
